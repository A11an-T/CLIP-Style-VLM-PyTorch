## Introduction to Tensors
### Creating Tensor 
Tensors are multidimension data 
PyTorch tensors are created using 'torch.Tensor()' https://docs.pytorch.org/docs/2.14/tensors.html (take notes)



```python
#scalar
scalar = torch.tensor(7)
scalar
```

```python
scalar.ndim #scalar has no dimension
```

```python
#Get tensor back as python int
scalar.item()
```

```python
#Vector - Magnitude and Direction
vector = torch.tensor([7,7])
vector
```

```python
vector.ndim
#why only 1? number of square brackets 
```

```python
torch.Size([2]) #element of 2
```

```python
#MATRIX - LINEAR ALGEBRA
matrix = torch.tensor([[7, 8], [9, 10]])
matrix
```

```python
matrix.ndim
```

```python
matrix.shape
#2 by 2 matrixs
```

```python
#Tensor
tensor = torch.tensor([[[1,2,3],[3,4,5],[3,2,1]]])
tensor
```

```python
tensor.ndim
```

```python
tensor.shape
```

```python
tensor[0]
```
